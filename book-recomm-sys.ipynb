### Popularity Based Recommendation System

```python
# highest avg rated books - we'll show - and here books for consideration are with minimum of 50 ratings
ratings_with_name = ratings.merge(books, on='ISBN')
ratings_with_name
```

```python
# avg rating + (total votes>50) : book_title basis groupby and each book have how many votes
num_rating_df = ratings_with_name.groupby('title').count()['rating'].reset_index()
num_rating_df.rename(columns={'rating': 'num_ratings'}, inplace=True)
num_rating_df
```

```python
avg_rating_df = ratings_with_name.groupby('title').mean(numeric_only=True)['rating'].reset_index()
avg_rating_df.rename(columns={'rating': 'avg_ratings'}, inplace=True)
avg_rating_df
```

```python
popular_df = num_rating_df.merge(avg_rating_df, on='title')
popular_df
```

```python
popular_df = popular_df[popular_df['num_ratings'] >= 50].sort_values('avg_ratings', ascending=False).head(50)
popular_df
```

```python
popular_df = popular_df.merge(books, on='title').drop_duplicates('title')[
    ['title', 'author', 'year', 'Image-URL-M', 'num_ratings', 'avg_ratings']
]
popular_df
```

### Collaborative Filtering Based Recommended System

```python
x = ratings_with_name.groupby('user_id')['rating'].count() > 200
trusted_users = x[x].index  # users who have rated more than 200 books
trusted_users
```

```python
filtered_rating = ratings_with_name[ratings_with_name['user_id'].isin(trusted_users)]
filtered_rating
# till now we have done filtering on the basis of users, now for books
```

```python
y = filtered_rating.groupby('title').count()['rating'] >= 50
famous_books = y[y].index
```

```python
final_ratings = filtered_rating[filtered_rating['title'].isin(famous_books)]
pt = final_ratings.pivot_table(index='title', columns='user_id', values='rating')
pt.fillna(0, inplace=True)
pt
```

```python
from scipy.sparse import csr_matrix
from sklearn.decomposition import TruncatedSVD
from sklearn.metrics.pairwise import cosine_similarity

sparse_pt=csr_matrix(pt.values)

# center each user's ratings around their own mean (only over books they rated)
col_sums=np.asarray(sparse_pt.sum(axis=0)).flatten()
col_counts=np.diff(sparse_pt.tocsc().indptr)
col_counts[col_counts==0]=1
user_means=col_sums/col_counts

centered=sparse_pt.tocsc().astype(float)
for col in range(centered.shape[1]):
    start,end=centered.indptr[col],centered.indptr[col+1]
    centered.data[start:end]-=user_means[col]
centered=centered.tocsr()

# down-weight users who rate almost everything (inverse user frequency)
iuf=np.clip(np.log(pt.shape[0]/col_counts),0.05,None)
weighted=centered.tocsc()
for col in range(weighted.shape[1]):
    start,end=weighted.indptr[col],weighted.indptr[col+1]
    weighted.data[start:end]*=iuf[col]
weighted=weighted.tocsr()

# reduce to latent factors before comparing books - cuts noise from sparse overlaps
svd=TruncatedSVD(n_components=min(50,min(weighted.shape)-1),random_state=42)
book_factors=svd.fit_transform(weighted)

similarity_scores=cosine_similarity(book_factors)
similarity_scores.shape
```

```python
# Keep this in sync with app.py's TOP_N. It's what decides how many
# similar books get shown, both here and on the site.
TOP_N = 5
```

```python
def recommend(book_name):
    # from book name fetch index
    index=np.where(pt.index==book_name)[0][0]
    similar_items=sorted(list(enumerate(similarity_scores[index])), key=lambda x: x[1], reverse=True)[1:TOP_N+1]
    data=[]
    for i in similar_items:
        item=[]
        temp_df=books[books['title']==pt.index[i[0]]]
        item.extend(list(temp_df.drop_duplicates('title')['title'].values))
        item.extend(list(temp_df.drop_duplicates('title')['author'].values))
        item.extend(list(temp_df.drop_duplicates('title')['Image-URL-M'].values))
        data.append(item)
    return data
```

```python
recommend('Message in a Bottle')
```
